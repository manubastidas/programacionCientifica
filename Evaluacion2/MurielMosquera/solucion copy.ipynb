## La capa Fourier

$W_1$ tiene $2K$ filas: las primeras $K$ son $\sin(2\pi k t)$ y las siguientes $K$ son $\cos(2\pi k t)$, para $k=1,\dots,K$.
Al multiplicar $W_1 x$ (antes de $\tanh$), cada fila calcula la proyección de $x$ sobre un seno o coseno de frecuencia $k$
— eso es, por definición, una transformada de Fourier en su versión seno/coseno (parte real/imaginaria de la DFT).

```python
def build_fourier_layer(K, t):
    """Construye W1 (2K, len(t)): filas sin(2*pi*k*t) seguidas de cos(2*pi*k*t), k=1..K."""
    ks = jnp.arange(1, K + 1)
    S = jnp.sin(2 * jnp.pi * jnp.outer(ks, t))
    C = jnp.cos(2 * jnp.pi * jnp.outer(ks, t))
    return jnp.concatenate([S, C], axis=0)

W1_fourier = build_fourier_layer(K, jnp.array(t))
print("W1_fourier:", W1_fourier.shape)
```

**Verificación contra `np.fft`.** `np.fft.rfft` recibe **una** señal 1-D (256 muestras reales) y devuelve un vector
complejo con las componentes de frecuencia $k=0,\dots,128$: la parte real es la proyección sobre cosenos, la parte
imaginaria (con signo opuesto) es la proyección sobre senos. Actúa señal por señal — para el dataset completo hay que
aplicarlo fila por fila de `X` (o con `axis=1`, que internamente sigue siendo "una FFT por señal"). Comparamos, para
una señal, las proyecciones manuales de $W_1$ contra `np.fft.rfft`:

```python
x0 = X[0]
proyeccion = W1_fourier @ jnp.array(x0)
fft_np = np.fft.rfft(x0)

ks = np.arange(1, K + 1)
cos_manual = np.array(proyeccion[K:])
sin_manual = np.array(proyeccion[:K])
cos_fft = fft_np[1:K+1].real
sin_fft = -fft_np[1:K+1].imag

fig, axes = plt.subplots(1, 2, figsize=(9, 3.3))
axes[0].plot(ks, cos_manual, "o-", color="tab:purple", label=r"$W_1 x$ (coseno)")
axes[0].plot(ks, cos_fft, ".-", color="k", label=r"\texttt{np.fft.rfft} (Re)")
axes[0].set_xlabel(r"$k$"); axes[0].set_ylabel(r"proyección")
axes[0].set_title(r"Componente coseno"); axes[0].legend(fontsize=8)

axes[1].plot(ks, sin_manual, "o-", color="tab:purple", label=r"$W_1 x$ (seno)")
axes[1].plot(ks, sin_fft, ".-", color="k", label=r"\texttt{np.fft.rfft} ($-$Im)")
axes[1].set_xlabel(r"$k$")
axes[1].set_title(r"Componente seno"); axes[1].legend(fontsize=8)

fig.suptitle(r"Capa Fourier en su inicialización $\equiv$ transformada de Fourier de $x_0$", y=1.03)
fig.tight_layout()
plt.show()

error_max = np.max(np.abs(cos_manual - cos_fft)) + np.max(np.abs(sin_manual - sin_fft))
print(f"Discrepancia máxima W1 vs np.fft: {error_max:.2e}  (debe ser ~0)")

energia_media = float((X**2).mean())
print(f"Energía media del dataset (X**2).mean() = {energia_media:.3f}")
```

## La red en JAX

$$\hat x = W_2\,\tanh(W_1 x), \qquad \mathcal{L}=\tfrac1N\|\hat x-x\|^2$$

Para que la comparación entre inicializaciones sea **justa**, la matriz aleatoria se genera con la misma escala típica
por entrada que la matriz de Fourier ($\mathbb{E}[w^2]=0.5$, igual que un seno o coseno promedio) — así la diferencia
de resultados se debe a la *estructura* de $W_1$, no simplemente a que una parta con números más grandes que la otra.

```python
def init_fourier(K, t):
    return build_fourier_layer(K, t)

def init_random(K, N_dim, key):
    # misma varianza por entrada que sin/cos: E[w^2] = 0.5
    return random.normal(key, (2 * K, N_dim)) * jnp.sqrt(0.5)

def init_W2(K, N_dim, key):
    return random.normal(key, (N_dim, 2 * K)) / jnp.sqrt(2 * K)

def forward(params, x):
    W1, W2 = params
    h = jnp.tanh(W1 @ x)
    return W2 @ h

forward_batch = vmap(forward, in_axes=(None, 0))

def loss_fn(params, Xb):
    Xhat = forward_batch(params, Xb)
    return jnp.mean((Xhat - Xb) ** 2)

@jit
def train_step(params, Xb, eta):
    W1, W2 = params
    g_W1, g_W2 = grad(loss_fn)(params, Xb)
    return (W1 - eta * g_W1, W2 - eta * g_W2)

def error_relativo(params, Xb):
    Xhat = forward_batch(params, Xb)
    return float(jnp.linalg.norm(Xhat - Xb) / jnp.linalg.norm(Xb))

def entrenar(params0, Xb, n_iter, eta):
    params = params0
    losses = np.zeros(n_iter)
    rel_errs = np.zeros(n_iter)
    for i in range(n_iter):
        losses[i] = float(loss_fn(params, Xb))
        rel_errs[i] = error_relativo(params, Xb)
        params = train_step(params, Xb, eta)
    return params, losses, rel_errs
```

**Separación entrenamiento / validación.** Con 15 señales por régimen usamos 12 para entrenar y **3 como
validación** (nunca vistas durante el entrenamiento). El error de reconstrucción que se reporta como resultado debe
venir de la validación — el de entrenamiento siempre es optimista. Entrenamos, **por régimen**, dos redes (una por
inicialización), para aislar el efecto de $W_1$.

```python
def split_train_val(X, y, regimen, n_val=3, seed=SEED):
    idx = np.where(y == regimen)[0]
    rng = np.random.default_rng(seed + regimen)
    idx = rng.permutation(idx)
    return X[idx[n_val:]], X[idx[:n_val]]

resultados = {}
k_w2 = key
for r in [0, 1, 2]:
    Xtr, Xval = split_train_val(X, y, r)
    Xtr_j, Xval_j = jnp.array(Xtr), jnp.array(Xval)

    k_w2, sub1, sub2 = random.split(k_w2, 3)
    W2_0 = init_W2(K, N, sub1)

    W1_f0 = init_fourier(K, jnp.array(t))
    params_f, lossF, relF = entrenar((W1_f0, W2_0), Xtr_j, N_ITER, ETA)

    W1_r0 = init_random(K, N, sub2)
    params_r, lossR, relR = entrenar((W1_r0, W2_0), Xtr_j, N_ITER, ETA)

    resultados[r] = dict(
        Xtr=Xtr, Xval=Xval,
        params_f=params_f, lossF=lossF, relF=relF,
        params_r=params_r, lossR=lossR, relR=relR,
        err_val_f=error_relativo(params_f, Xval_j),
        err_val_r=error_relativo(params_r, Xval_j),
        err_tr_f=relF[-1], err_tr_r=relR[-1],
    )
    print(f"Régimen {r}: listo (Fourier y aleatoria, {N_ITER} iteraciones)")
```

## Compresión

Cada fila de $W_1^{Fourier}$, normalizada, es (casi) ortonormal frente a las demás sobre las 256 muestras: proyectar
$x$ sobre esas $2K=50$ direcciones da $2K$ coeficientes independientes. Ordenamos esos coeficientes por energía
($c_i^2$) y medimos qué fracción de la **energía total de la señal** ($\|x\|^2$) capturan los $m$ coeficientes más
grandes, para distintos $m$.

```python
row_norm = jnp.sqrt(jnp.sum(W1_fourier ** 2, axis=1, keepdims=True))
Phi = W1_fourier / row_norm

def curva_compresion(x, Phi):
    c = np.array(Phi @ jnp.array(x))
    orden = np.argsort(-np.abs(c))
    E_total = np.sum(x ** 2)
    errores, energia_acum = [], []
    for m in range(1, len(c) + 1):
        idx = orden[:m]
        x_aprox = (c[idx][:, None] * np.array(Phi)[idx]).sum(axis=0)
        errores.append(np.linalg.norm(x_aprox - x) / np.linalg.norm(x))
        energia_acum.append(np.sum(c[idx] ** 2) / E_total)
    return np.array(errores), np.array(energia_acum)

fig, ax = plt.subplots(figsize=(8, 5))
m_para_umbral = {}
for r in [0, 1, 2]:
    Xr = X[y == r]
    curvas = [curva_compresion(x, Phi) for x in Xr]
    err_prom = np.mean([c[0] for c in curvas], axis=0)
    ener_prom = np.mean([c[1] for c in curvas], axis=0)
    m_coefs = np.arange(1, 2 * K + 1)
    ax.plot(m_coefs, err_prom, "o-", ms=4, color=COLOR_REGIMEN[r], label=rf"régimen {r}")

    reach = np.where(ener_prom >= ENERGIA)[0]
    m_para_umbral[r] = int(reach[0]) + 1 if len(reach) > 0 else None

ax.axhline(np.sqrt(1 - ENERGIA), color="k", ls="--", lw=1,
           label=rf"error para ${int(ENERGIA*100)}\%$ de energía")
ax.set_xlabel(r"número de coeficientes conservados (de $2K=" + str(2*K) + r"$)")
ax.set_ylabel(r"error de reconstrucción relativo (promedio por régimen)")
ax.set_title(rf"Compresión con la capa Fourier ($K={K}$) --- umbral ${int(ENERGIA*100)}\%$ de energía", fontsize=14)
ax.grid(True, color="lightgray", linestyle="--", linewidth=0.6, alpha=0.7, zorder=0)
ax.legend(fontsize=10)
fig.tight_layout()
plt.savefig("fig_compresion.pdf")
plt.show()

print(f"Coeficientes necesarios para capturar el {int(ENERGIA*100)}% de la energía, por régimen:")
for r in [0, 1, 2]:
    m = m_para_umbral[r]
    print(f"  Régimen {r}: {m if m is not None else 'NO se alcanza con los 50 disponibles'}")
```

**Confirmación con los datos:** los resultados obtenidos son $m(95\%)=5$ para el régimen 0, $m(95\%)=8$ para el
régimen 1, y $m(95\%)=8$ para el régimen 2 (de $2K=50$ coeficientes disponibles).

Esto **confirma parcialmente** la hipótesis: el régimen 0 ($\mu=0$, sin dispersión) necesita claramente menos
coeficientes que los otros dos, consistente con que sus frecuencias son las más bajas y caen cómodamente dentro del
rango cubierto por $K=25$. Sin embargo, los regímenes 1 y 2 requieren el **mismo** número de coeficientes (8), en vez
de que el régimen 2 necesite estrictamente más —como se esperaría si su mayor $\mu$ empujara las frecuencias mucho
más allá del rango de $K=25$—.

Una explicación razonable es que, con $K=25$, el rango de frecuencias cubierto por la capa Fourier ya es lo
suficientemente amplio como para contener los modos de ambos regímenes sin problema —a diferencia de un $K$ más
pequeño, donde el régimen de mayor $\mu$ sí quedaría claramente en desventaja—. Es decir, la limitación que predice la
física (frecuencias que se salen del rango cubierto) existe, pero con este valor de $K$ no es lo bastante severa como
para separar claramente los regímenes 1 y 2 en términos de compresión.

  0.42–0.45 con Fourier) — ajusta mejor las 12 señales que efectivamente vio durante el entrenamiento.
- **En error de *validación*** —la métrica que realmente importa—, el patrón se invierte: Fourier generaliza mejor en
  los tres regímenes (0.83–1.16 frente a 1.01–1.35 con aleatoria).
- En el régimen foco (1), la curva de pérdida muestra un **cruce**: Fourier arranca más rápido (ventaja inicial por
  partir ya alineada con las frecuencias reales de la señal), pero la aleatoria la alcanza y termina con menor error
  de entrenamiento hacia la iteración 3000.
- Con $\eta=0.02$ y 3000 iteraciones, **ninguna** inicialización alcanza el umbral de error relativo $0.1$ en ningún
  régimen (ambas terminan en 0.33–0.45); las curvas de pérdida todavía tenían pendiente negativa al final, indicio de
  que el entrenamiento no llegó a convergencia con este presupuesto de iteraciones.

**Conclusión:** con datos tan escasos (12 señales de entrenamiento por régimen frente a ~25,600 parámetros), ambas
inicializaciones sobreajustan severamente. Arrancar en Fourier sí ayuda —pero el beneficio aparece en la capacidad de
generalización, no en la velocidad ni calidad de ajuste sobre los datos de entrenamiento. Un vistazo únicamente al
error de entrenamiento habría llevado a la conclusión contraria e incorrecta.

**3. ¿Cuánta información vive en pocos coeficientes de la capa Fourier?**

Bastante: con solo 5 coeficientes (de 50 disponibles) se captura el 95% de la energía en el régimen 0 (sin
dispersión), y con 8 coeficientes se logra lo mismo en los regímenes 1 y 2. Esto es consistente con que las señales
son, por construcción, sumas de solo 3 modos —la mayoría de las direcciones de la capa Fourier no contribuyen energía
significativa a estas señales en particular.
